```python
def text_clean(text, LETTERS='ABCDEFGHIJKLMNOPQRSTUVWXYZ'):
    text = text.upper()
    clean_text = ''
    
    for char in text:
        if char in LETTERS:
            clean_text += char

    return clean_text
```

```python
def text_block(text, n=5):
    blocked_text = ''
    for char in text:
        blocked_text += char
        if len(blocked_text.replace(' ', '')) % n == 0:
            blocked_text += ' '

    return blocked_text.strip()
```

```python
def eea(k, m, verbose=False):
    a, b = m, k
    x0, x1 = 1, 0
    y0, y1 = 0, 1

    # a = x0 * m + y0 * k
    # b = x1 * m + y1 * k
    if verbose:
        print(f"{a} = \t{m}({x0}) \t+ {k}({y0})")
    
    while b != 0:
        q = a // b # Determine the number of multiples of b to subtract from a
        a, x0, y0, b, x1, y1 = b, x1, y1, a - q*b, x0 - q*x1, y0 - q*y1 
        if verbose:
            print(f"{a} = \t{m}({x0}) \t+ {k}({y0})")
    
    if verbose:
        print(f"{b} = \t{m}({x1}) \t+ {k}({y1})")
    
    if a != 1:
        raise ValueError(f"Modular inverse does not exist for k = {k} and m = {m} (gcd = {a}).")

    return y0 % m
```

```python
def affine(text, k_a, k_m, decrypt=False, LETTERS = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'):
    cleaned_text = text_clean(text)
    final_text = ''
    k_m_inv = eea(k_m, len(LETTERS))

    for char in cleaned_text:
        text_numerical = LETTERS.find( char )
        if decrypt:
            new_text_numerical = ((text_numerical - k_a) * k_m_inv)  % len(LETTERS)
        else:
            new_text_numerical = (k_m * text_numerical + k_a) % len(LETTERS)
        
        final_text += LETTERS[new_text_numerical]

    if decrypt:
        return final_text.lower()
    else:
        return text_block(final_text)
```

```python
with open('activity10_plaintext.txt') as f:
    plaintext = f.read()
```

```python
affine_ciphertext = affine(plaintext, 20, 19)
```

```python
with open('activity10_affine-ciphertext.txt', 'w') as f:
    print(affine_ciphertext, file=f, end='')
```

```python
caesar_ciphertext = affine(plaintext, 17, 1)
```

```python
with open('activity10_caesar-ciphertext.txt', 'w') as f:
    print(caesar_ciphertext, file=f, end='')
```
